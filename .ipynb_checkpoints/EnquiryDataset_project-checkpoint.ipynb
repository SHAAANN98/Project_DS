# Data Visualization


## Personal information

### Top 10 Locations

```python
top10 = (
    data["location"]
      .value_counts()    
      .head(10)     
      .reset_index()
)

top10.columns = ["location", "enquiry_count"]

top10
```

```python
import plotly.express as px

fig = px.bar(
    top10,
    x="enquiry_count",
    y="location",
    orientation="h",
    text="enquiry_count",
)

fig.update_layout(
    title="Top 10 Locations by Student Enquiries",
    yaxis=dict(autorange="reversed"),
    height=500
)

fig.update_traces(textposition="outside")

fig.show()
```

- #### Kozhikode is the location leading with highest Enquiry.
- #### Focus marketing or outreach efforts on these key locations to increase conversions.



### Qualification wise student count

```python
qualification_count = (
    data["qualification"]
      .value_counts()
      .reset_index()
)

qualification_count.columns = ["qualification", "student_count"]

qualification_count
```

```python
import plotly.express as px

fig = px.bar(
    qualification_count,
    x="qualification",
    y="student_count",
    text="student_count",
)

fig.update_layout(
    title="Qualification-wise Student Enquiry Count",
    height=500
)

fig.update_traces(textposition="outside")

fig.show()
```

### Qualification-wise student count by status

- #### Monitoring status by qualification helps target follow-ups and improve enrolment.

###  Monthwise trend line of each year

```python
data['date'] = pd.to_datetime(data['date'], dayfirst=True, errors='coerce')

data['year'] = data['date'].dt.year
data['month'] = data['date'].dt.month
```

```python
month_year_trend = (
    data.groupby(['year', 'month'])
      .size()
      .reset_index(name='count')
      .sort_values(['year', 'month'])
)
```

```python
import calendar
month_year_trend['month_name'] = month_year_trend['month'].apply(lambda x: calendar.month_name[x])
```

```python
fig = px.line(month_year_trend,
              x='month_name',
              y='count',
              color='year',
              markers=True,
              title='Month-wise Enquiry Trend by Year')

month_order = [
    'January','February','March','April','May','June',
    'July','August','September','October','November','December'
]

fig.update_xaxes(categoryorder='array', categoryarray=month_order)

fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Number of Enquiries'
)

fig.show()
```

- #### Enquiry activity fluctuates across months within each year.
- #### The pattern suggests seasonal variations in student interest.
- #### Identifying peak months can help institutions plan marketing and admission campaigns more effectively.

### Top technologies per year trend

```python
tech_year = (
    data.groupby(['year', 'technology'])
      .size()
      .reset_index(name='count')
)

tech_year
```

```python
top_tech = (
    data['technology']
    .value_counts()
    .head(5)
    .index
)

tech_year_top = tech_year[tech_year['technology'].isin(top_tech)]
```

```python
fig = px.line(tech_year_top,
              x='year',
              y='count',
              color='technology',
              markers=True,
              title='Top Technologies Trend by Year')

fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Number of Students'
)

fig.show()
```

- #### Demand for different technologies varies across years.
- #### python show a steady increase in student interest over time.
- #### The trend reflects how evolving industry demand influences student technology preferences.
